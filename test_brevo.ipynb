```python
import os
import requests

def parse_env(file_path):
    env_vars = {}
    with open(file_path, "r") as f:
        for line in f:
            if line.strip() and not line.startswith("#"):
                key, *val = line.strip().split("=", 1)
                if val:
                    env_vars[key.strip()] = val[0].strip().strip('"').strip("'")
    return env_vars

env_vars = parse_env("d:/SHADOW SHOP/.env")
brevo_key = env_vars.get("BREVO_API_KEY", "").strip()
email_from = env_vars.get("EMAIL_FROM", "test@shadowshop.local")

print(f"Key loaded: {len(brevo_key)} characters")

payload = {
    "sender": {"name": "SHADOW SHOP Test", "email": email_from},
    "to": [{"email": email_from}],
    "subject": "SHADOW SHOP Email System Test - Brevo API",
    "htmlContent": "<h2>Success!</h2><p>The SHADOW SHOP email system is now successfully using the Brevo HTTPS Transactional API.</p>"
}

try:
    res = requests.post(
        "https://api.brevo.com/v3/smtp/email",
        json=payload,
        headers={
            "api-key": brevo_key,
            "accept": "application/json",
            "content-type": "application/json"
        }
    )
    print("Status:", res.status_code)
    try:
        print("Response:", res.json())
    except:
        print("Response:", res.text)
except Exception as e:
    print("Error:", e)
```
